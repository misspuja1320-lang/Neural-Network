```python
# implement McCulloch-Pitts OR operation neural model
def mcp_neuron(x1,x2,threshold):
    weighted_sum=x1+x2
    if weighted_sum>=threshold:
        return 1
    else:
        return 0
threshold=1
print("x1 x2 output")
inputs=[(0,0),(0,1),(1,0),(1,1)]
for x1,x2 in inputs:
    output=mcp_neuron(x1,x2,threshold)
    print(x1,"",x2,"",output)
```

```python
# implement McCulloch-Pitts AND-OR operation neural model
def mcp_neuron(inputs,weights,threshold):
    weighted_sum=0
    for i in range(len(inputs)):
        weighted_sum+=inputs[i]*weights[i]
    if weighted_sum>=threshold:
        return 1
    else:
        return 0
input_data=[[0,0],[0,1],[1,0],[1,1]]
weights=[1,1]
print("AND gate")
print("x1 x2 output")
for inputs in input_data:
    output=mcp_neuron(inputs,weights,2)
    print(inputs[0],inputs[1],output)
print("-------------")
print("OR gate")
print("x1 x2 output")
for inputs in input_data:
    output=mcp_neuron(inputs,weights,1)
    print(inputs[0],inputs[1],output)
```
