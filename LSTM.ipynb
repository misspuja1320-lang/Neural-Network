```python
# Write a Python program using TensorFlow/Keras to implement an LSTM (Long Short-Term Memory) neural network for sequence prediction. 
```

```python
import numpy as np 
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import LSTM, Dense 
from tensorflow.keras.callbacks import EarlyStopping 
```

```python
data = np.arange(0, 1000, 10, dtype=float) 
data = data / 1000.0
x, y = [], [] 
for i in range(len(data) - 3): 
    x.append(data[i:i+3]) 
    y.append(data[i+3]) 
x = np.array(x).reshape(-1, 3, 1) 
y = np.array(y) 
model = Sequential([LSTM(32, input_shape=(3, 1)), Dense(1)]) 
model.compile(optimizer='adam', loss='mse') 
model.fit(x, y,epochs=500,batch_size=8,verbose=0, callbacks=[EarlyStopping(monitor='loss',patience=30, restore_best_weights=True)])
test = np.array([60, 70, 80], dtype=float) 
test = (test / 1000.0).reshape(1, 3, 1) 
prediction = model.predict(test, verbose=0)[0, 0] * 1000
print("Input sequence: 60, 70, 80") 
print("Predicted next number:", round(prediction, 2)) 
```
