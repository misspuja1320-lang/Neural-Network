```python
# Digit prediction using CNN for MNIST dataset
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
from tensorflow.keras.datasets import mnist
```

```python
(x_train,y_train),(x_test,y_test)=mnist.load_data()
print("Training data shape:",x_train.shape)
print("Tesing data shape:",x_test.shape)
print("Training data shape:",y_train.shape)
print("Tesing data shape:",y_test.shape)
x_train=x_train.astype("float32")/255.00
x_test=x_test.astype("float32")/255.00
model=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation='relu'),
    Dense(64,activation='relu'),
    Dense(10,activation='softmax')])
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=['accuracy'])
model.summary()
history=model.fit(x_train,y_train,batch_size=64,epochs=5,validation_split=0.1)
test_loss,test_acc=model.evaluate(x_test,y_test,verbose=2)
print("Test loss:",test_loss)
print("Test accuracy:",test_acc)
predictions=model.predict(x_test)
predicted_digit=np.argmax(predictions[2])
print("Actual digit:",y_test[2])
print("Predicted digit:",predicted_digit)
```
